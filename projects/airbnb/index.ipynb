````python
#| label: setup
#| include: false
from pathlib import Path
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Put the CSV next to this QMD and name it "zurich_listings.csv".
# The fallbacks make the page easier to preview while you are reorganizing the project.
candidates = [
    Path("zurich_listings.csv"),
    Path("zurich_listings (1).csv"),
    Path("projects/airbnb/zurich_listings.csv"),
    Path("projects/airbnb/zurich_listings.csv"),
    Path("projects/airbnb/zurich_listings.csv"),
    Path("projects/airbnb/zurich_listings.csv")
]
DATA_PATH = next((p for p in candidates if p.exists()), None)
if DATA_PATH is None:
    raise FileNotFoundError("Place zurich_listings.csv in the same folder as this QMD, or update DATA_PATH in the setup chunk.")

zurich_df = pd.read_csv(DATA_PATH)
zurich_df["price"] = pd.to_numeric(zurich_df["price"].astype(str).str.replace(r"[^\d.]", "", regex=True), errors="coerce")

# Minimal cleaning used by the selected case-study analyses.
if "bathrooms" in zurich_df.columns and zurich_df["bathrooms"].isna().any() and "bathrooms_text" in zurich_df.columns:
    parsed_bath = pd.to_numeric(zurich_df["bathrooms_text"].astype(str).str.extract(r"([0-9]+(?:\.[0-9]+)?)")[0], errors="coerce")
    zurich_df["bathrooms"] = zurich_df["bathrooms"].fillna(parsed_bath)

zurich_df["beds"] = zurich_df["beds"].fillna(zurich_df["accommodates"].apply(lambda x: max(1, round(x / 2)) if pd.notna(x) else np.nan))
zurich_df.loc[zurich_df["bedrooms"].isna() & (zurich_df["accommodates"] <= 2), "bedrooms"] = 1
zurich_df.loc[zurich_df["bedrooms"].isna() & zurich_df["accommodates"].between(3,4), "bedrooms"] = 2
zurich_df.loc[zurich_df["bedrooms"].isna() & (zurich_df["accommodates"] > 4), "bedrooms"] = 3
zurich_df["bedrooms"] = zurich_df["bedrooms"].fillna((zurich_df["beds"] / 1.5).round())

zurich_df["host_is_superhost"] = zurich_df["host_is_superhost"].map({"t":1,"f":0}).fillna(0)
zurich_df["amenities_clean"] = zurich_df["amenities"].fillna("").str.replace(r'[\[\]"{}]', '', regex=True).str.lower()
zurich_df["has_parking"] = zurich_df["amenities_clean"].str.contains(r"\bparking\b|\bfree parking\b", na=False).astype(int)

top_neighborhood = zurich_df["neighbourhood_cleansed"].value_counts().head(10)
top_neigh_list = top_neighborhood.index
df_top = zurich_df[zurich_df["neighbourhood_cleansed"].isin(top_neigh_list)].copy()

CORAL = "#FF5A5F"; BLACK = "#17191F"; GRAY = "#687182"; LIGHT = "#F6E7E9"
def finish(fig, height=430):
    fig.update_layout(template="plotly_white", font=dict(family="Inter, Arial", size=11, color=BLACK), height=height, margin=dict(l=55,r=25,t=65,b=60), paper_bgcolor="white", plot_bgcolor="white")
    return fig
````

```{=html}
<section id="clustering" class="case-section"><p class="eyebrow"><span class="num">06</span> CLUSTERING ANALYSIS</p><h2>Five segments reveal different pricing and demand strategies.</h2><p class="wide-copy">I clustered listings using price, accommodates, number of reviews, review score and an engineered 30-day booking-rate proxy. Features were standardized before K-Means so variables measured on larger scales would not dominate the distance calculation. The elbow curve did not have a sharp break, but improvement began to diminish around five clusters, so k = 5 was selected as an interpretable segmentation.</p></section>
```

````python
#| label: clustering-dashboard
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
cl=zurich_df.copy();cl['booked_30']=1-cl['availability_30']/30;cl['review_scores_rating']=cl['review_scores_rating'].replace(0,np.nan);cl=cl.fillna(cl.median(numeric_only=True))
features=['price','accommodates','number_of_reviews','review_scores_rating','booked_30']
X=StandardScaler().fit_transform(cl[features])
inertia=[]
for k in range(2,10): inertia.append(KMeans(n_clusters=k,random_state=42).fit(X).inertia_)
km=KMeans(n_clusters=5,random_state=42).fit(X);cl['cluster']=km.labels_
names={0:'Budget High-Occupancy',1:'High-Price Low-Demand',2:'High-Reputation High-Demand',3:'Premium Large-Capacity',4:'Lower-Rated Mid-Range'}
colors={0:'#FF5A5F',1:'#27B8D0',2:'#FF9CA3',3:'#17191F',4:'#B8B8B8'}
cl['cluster_name']=cl['cluster'].map(names)
sumdf=cl.groupby('cluster').agg(avg_price=('price','mean'),avg_booked=('booked_30','mean'),avg_accommodates=('accommodates','mean'),listing_count=('cluster','size'),reviews=('number_of_reviews','mean'),rating=('review_scores_rating','mean')).reset_index()
sumdf['booked_pct']=sumdf['avg_booked']*100
fig=make_subplots(rows=2,cols=2,subplot_titles=('1. Elbow Method','2. Average Price & Demand Profile','3. Number of Listings by Cluster','4. Average Price by Cluster'),vertical_spacing=.22,horizontal_spacing=.15)
fig.add_trace(go.Scatter(x=list(range(2,10)),y=inertia,mode='lines+markers',line=dict(color=CORAL,width=3),marker=dict(size=7),showlegend=False),1,1)
for _,r in sumdf.iterrows(): fig.add_trace(go.Scatter(x=[r.avg_price],y=[r.booked_pct],mode='markers+text',text=[names[r.cluster]],textposition='top center',marker=dict(size=max(13,min(42,r.listing_count/30)),color=colors[r.cluster],opacity=.82),name=names[r.cluster],showlegend=False,hovertemplate=f"{names[r.cluster]}<br>Avg price: CHF %{{x:.2f}}<br>30-day booked: %{{y:.1f}}%<br>Listings: {int(r.listing_count)}<extra></extra>"),1,2)
fig.add_trace(go.Bar(x=[names[c] for c in sumdf.cluster],y=sumdf.listing_count,marker_color=[colors[c] for c in sumdf.cluster],text=sumdf.listing_count,textposition='outside',showlegend=False),2,1)
fig.add_trace(go.Bar(x=[names[c] for c in sumdf.cluster],y=sumdf.avg_price,marker_color=[colors[c] for c in sumdf.cluster],text=[f"{v:.2f}" for v in sumdf.avg_price],textposition='outside',showlegend=False),2,2)
fig.update_xaxes(title_text='Number of Clusters (k)',row=1,col=1);fig.update_yaxes(title_text='Inertia',row=1,col=1)
fig.update_xaxes(title_text='Average Price (CHF)',row=1,col=2);fig.update_yaxes(title_text='Booked in Next 30 Days (%)',row=1,col=2)
fig.update_yaxes(title_text='Number of Listings',row=2,col=1);fig.update_yaxes(title_text='Average Price (CHF)',row=2,col=2)
fig.update_xaxes(tickangle=-28,row=2,col=1);fig.update_xaxes(tickangle=-28,row=2,col=2)
finish(fig,760).update_layout(title=dict(text='Zurich Airbnb Listing Segments',x=.02),margin=dict(l=65,r=35,t=90,b=100));fig.show()
````
