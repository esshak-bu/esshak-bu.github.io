from pyspark.sql.functions import regexp_replace, split, explode, trim, lower, col, lit

# clean the string and remove unwanted characters such as list brackets, quotes, and new lines
def extract_skills(dataframe, skill_col, skill_type):
    cleaned = dataframe.withColumn(
        "SKILL_CLEAN",
        regexp_replace(col(skill_col), r'[\[\]\n"]', "")
    )

# explode so its one row per skill and rename column
    exploded = cleaned.select(
        "ROLE_CATEGORY",
        explode(split(col("SKILL_CLEAN"), ",")).alias("SKILL")
    )

# clean formatting so everything is lower, trim any spaces, and remove empty values
    return exploded.withColumn(
        "SKILL",
        trim(lower(col("SKILL")))
    ).filter(
        (col("SKILL").isNotNull()) &amp; (col("SKILL") != "")
    ).withColumn(
        "SKILL_TYPE",
        lit(skill_type)
    )</code></pre>

from pyspark.sql.window import Window
from pyspark.sql.functions import col, row_number

def get_top_skills_by_role(df):
    counts = df.groupBy("ROLE_CATEGORY", "SKILL").count()

    window = Window.partitionBy("ROLE_CATEGORY").orderBy(col("count").desc())

    ranked = counts.withColumn(
        "rank",
        row_number().over(window)
    ).filter(col("rank") &lt;= 15)

    return ranked</code></pre>

````python
#| echo: false
#| warning: false

import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots

PORTFOLIO_FONT = "Inter, Helvetica Neue, Arial, sans-serif"
TEXT = "#223047"
MUTED = "#718096"
GRID = "#F1F3F6"

ROLE_COLORS = {
    "Data Analyst": "#7FB3E6",
    "Data Scientist": "#A98BE3",
    "AI Engineer/Analyst": "#6BCB9A"
}

def apply_portfolio_style(fig, height=None):
    fig.update_layout(
        template="plotly_white",
        paper_bgcolor="rgba(0,0,0,0)",
        plot_bgcolor="rgba(0,0,0,0)",
        font=dict(family=PORTFOLIO_FONT, color=TEXT, size=12),
        title_font=dict(family=PORTFOLIO_FONT, color=TEXT, size=18),
        margin=dict(l=24, r=45, t=76, b=52),
        hoverlabel=dict(
            bgcolor="white",
            bordercolor="#E4E7EC",
            font=dict(family=PORTFOLIO_FONT, color=TEXT)
        )
    )
    if height:
        fig.update_layout(height=height)

    fig.update_xaxes(
        showgrid=True,
        gridcolor=GRID,
        zeroline=False,
        tickfont=dict(family=PORTFOLIO_FONT, color=MUTED, size=11),
        title_font=dict(family=PORTFOLIO_FONT, color=MUTED, size=12),
        automargin=True
    )
    fig.update_yaxes(
        showgrid=False,
        zeroline=False,
        tickfont=dict(family=PORTFOLIO_FONT, color=TEXT, size=11),
        automargin=True
    )
    return fig

def role_skill_figure(data, title, top_n=15):
    roles = ["Data Analyst", "Data Scientist", "AI Engineer/Analyst"]

    fig = make_subplots(
        rows=3,
        cols=1,
        subplot_titles=roles,
        vertical_spacing=0.075
    )

    for row_num, role in enumerate(roles, start=1):
        rows = sorted(data[role], key=lambda x: x[1], reverse=True)[:top_n]
        rows = rows[::-1]

        fig.add_trace(
            go.Bar(
                x=[v for _, v in rows],
                y=[k for k, _ in rows],
                orientation="h",
                text=[f"{v:,}" for _, v in rows],
                textposition="outside",
                cliponaxis=False,
                marker=dict(
                    color=ROLE_COLORS[role],
                    line=dict(width=0)
                ),
                hovertemplate="<b>%{y}</b><br>%{x:,} postings<extra></extra>",
                showlegend=False
            ),
            row=row_num,
            col=1
        )

        fig.update_xaxes(
            title_text="Number of job postings",
            row=row_num,
            col=1
        )

    apply_portfolio_style(fig, height=1220)
    fig.update_layout(
        title=dict(
            text=title,
            x=0.01,
            xanchor="left"
        ),
        bargap=0.28
    )

    # Make the three role headings match their bar colors.
    for annotation, role in zip(fig.layout.annotations, roles):
        annotation.update(
            font=dict(
                family=PORTFOLIO_FONT,
                size=15,
                color=ROLE_COLORS[role]
            ),
            x=0,
            xanchor="left"
        )

    return fig
````

      <div>
        <h3>Clustering is exploratory, not ground truth.</h3>
        <p>
          K-Means assumes distance-based, relatively compact clusters. The gradual
          WSSSE curve and PCA overlap indicate that this labor-market data does not
          naturally divide into sharply separated groups. PCA also compresses the
          full feature space into two dimensions, so some structure can be lost in
          visualization.

          silhouette scores, compare cluster membership against the original role
          categories, and evaluate cluster stability across several random seeds.
